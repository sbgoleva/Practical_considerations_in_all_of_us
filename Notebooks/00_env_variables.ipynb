```r
library(jsonlite)

wb <- function(...) fromJSON(system(paste("wb", ..., "--format=json"), intern = TRUE))

workspace <- wb("workspace describe")
GOOGLE_CLOUD_PROJECT <- workspace$googleProjectId

resources <- wb("resource list")

# WORKSPACE_BUCKET
bucket_row <- resources[resources$resourceType == "GCS_BUCKET" &
                        grepl("practical_considerations_bucket", resources$id) &
                        !grepl("temporary", resources$id), ]
WORKSPACE_BUCKET <- paste0("gs://", bucket_row$bucketName[1])

# WORKSPACE_CDR
bq <- resources[resources$resourceType %in% c("BQ_DATASET", "BIGQUERY_DATASET"), ]
cdr_row <- bq[grepl("^C\\d{4}Q\\d+R\\d+$", bq$datasetId), ]
WORKSPACE_CDR <- paste0(cdr_row$projectId[1], ".", cdr_row$datasetId[1])
```
